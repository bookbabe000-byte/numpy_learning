```python
language=["c++","python","c#","java","rust"]
popularity=[56,89,67,60,34]
plt.pie(popularity)
plt.show()
```

```python
language=["c++","python","c#","java","rust"]
popularity=[56,89,67,60,34]
plt.pie(popularity,labels=language)
plt.show()
```

```python
language=["c++","python","c#","java","rust"]
popularity=[56,89,67,60,34]
plt.pie(popularity,labels=language,autopct="%2f%%")
plt.show()
```

```python
language=["c++","python","c#","java","rust"]
popularity=[56,89,67,60,34]

explode=[0.5,0.5,0.5,0.5,0.5]
plt.pie(popularity,labels=language,autopct="%2f%%",explode=explode)
plt.show()
```
